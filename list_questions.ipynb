```python
# find largest value in a list
nums=[11,22,-1,3,-44,-55,9,87,-54,22,-7,-71,1,1,-2]
largest=nums[0]

for n in nums:
    if n>largest:
        largest=n
print("Largest no is ",largest)
```

```python
# generate a list having unique numbers from another list having duplicates
nums=[11,22,-1,3,-44,-55,9,87,-54,22,-7,-71,1,1,-2,-2,11,11,22,3,-44]
unique=[]

for n in nums:
    if n not in unique:
        unique.append(n)

print('Unique list is ',unique)


```

# LIST COMPREHENSION
LIST COMPREHENSION is a modern days functional programming where
we can avoid writing any LOOPS and if condtion in the process.
So LIST COMPREHENSION is nothing but generating a new list
from an existing list but no for loop while loop if if else will be applied.
So we can derive a new list from the old given list in a single line only
as we are avoiding for loop , while loop and also if,else logic

Syntax is :-

target=[  apply logic on iteration_variable  for iteration_variable in old_list  if .....condition.....]


```python
nums=[1,2,3,4,5,6,7,8,9]

result=[ _+100   for _ in nums ] #LIST COMPREHENSION

print(result)
```
