```python
def process(mylist:list):
    """
    this is a function
    written by Amitava
    developed on Wednesday
    """
    maxnum=max(mylist)
    minnum=min(mylist)
    return maxnum,minnum

nums=[22,33,44,55,66]
t1,t2=process(nums) # UNPACKING
print('max :-',t1)
print('min :-',t2)

```

```python
'''
WAF that takes a list of lists where each list contains(A SHOPPING CART):
item_name
quantity
price_per_item
The function must return the total cost of all items:--
'''

def total_cost_cart(cart:list):
    total=[qty * price for _,qty,price in cart]
    return total

cart_items=[
    ['apple',3,100],
    ['banana',5,30],
    ['milk',2,120]
]

print('total cart value is ',total_cost_cart(cart_items))


```
